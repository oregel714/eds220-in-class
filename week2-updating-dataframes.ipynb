# 4. Updating Dataframes

```python
# Adding a single column 
import numpy as np
import pandas as pd 

# Create a random number generator with a seed 
rng = np.random.default_rng(seed=42)

# Import data
URL = 'https://raw.githubusercontent.com/allisonhorst/palmerpenguins/main/inst/extdata/penguins.csv'
penguins = pd.read_csv(URL)
```

### Using dictionary-like syntax: 
Simple syntax to add a new column to a `pandas.DataFrame` is: 

`df['new_col_name'] = new_column_values`

Adding a new key-value pair to a dictionary: 

`dict[new_key] = new_value`

```python
# Adding a new column body_mass_kg
penguins['body_mass_kg'] = penguins['body_mass_g'] / 1000

# Confirm the new column is in the data frame 
print("body_mass_kg is in the dataframe's columns: ", 'body_mass_kg' in penguins.columns)

# Look at the new column 
penguins.head()
```
