```python
import pandas as pd 

df = pd.read_csv('data/wetlands_seasonal_bird_diversity.csv')

df.head()
```

```python
df.plot()
```

```python
df.plot(x = 'year', 
        y = 'CSM_winter',
        title = 'Bird species registered during winter at Carpenteria Salt Marsh',
        xlabel = 'Year',
        ylabel = 'Number of bird species',
        color = 'green',
        legend = False
       )
```

```python
df.plot(x = 'year', 
        y = 'MUL_spring',
        title = 'Bird species registered during spring at Mugu Lagoon',
        xlabel = 'Year',
        ylabel = 'Number of bird species',
        color = 'green',
        legend = False
       )
```

```python
df.loc[df['MUL_spring'].isna()]
```

```python
df.plot(x = 'year',
       y = ['TJE_spring', 'TJE_fall'],
       title = 'Seasonal bird surveys at Tijuana Estuary',
        xlabel = 'Year',
        ylabel = 'Number of bird species',
        color = {'TJE_spring' : '#F48FB1',
                 'TJE_fall' : '#AB47BC'},
        subplots = True
       );
```

```python
# Set `column_name` column in df as the new index (reassignment)
df = df.set_index('column_name')
```

```python
# Set `column_name` column in df as the new index (modify df in-place)
df.set_index('column_name', inplace=True)
```

```python
df = df.set_index('year')
df.head()
```

```python
df.plot(y = 'CSM_winter')
```

```python
df = df.reset_index()
df.head()
```

```python
df.set_index('year').loc[:, 'SDW_winter': 'TJE_fall'].plot()
# We are setting the dataframe's index to years (x-axis) and plotting the bird species on the y-axis
# using .loc and selecting the species from 'sdw_winter' to 'TJE_fall' while keeping the row indexer blank thus selecting all the rows. .
```

```python
# Read in data
URL = 'https://raw.githubusercontent.com/allisonhorst/palmerpenguins/main/inst/extdata/penguins.csv'
penguins = pd.read_csv(URL)

penguins.head()
```

```python
penguins.info()
```

```python
penguins.describe()
```

```python
penguins[['species', 'sex', 'island', 'year']].nunique()
```

```python
penguins['species'].unique()
```

```python
penguins['species'].value_counts()
```

```python
penguins.plot(kind = 'scatter',
    x = 'flipper_length_mm',
     y = 'body_mass_g',
    title = 'Flipper length and body mass for Palmer penguins',
    xlabel = 'Flipper length (mm)',
    ylabel = 'Body mass(g)',
    color = '#ff3b01',
    alpha = 0.4 # controls transparency
    )
```

```python
smallest = penguins['body_mass_g'].nsmallest(10)
smallest
```

```python
smallest.plot(kind = 'bar')
```

```python
penguins.nsmallest(10, 'body_mass_g')
```

```python
penguins.plot(kind = 'hist')
```

```python
penguins['flipper_length_mm'].plot(kind = 'hist',
                                  title = 'Penguin flipper length (mm)',
                                  xlabel = 'Flipper length (mm)',
                                  grid = True)
```

```python
penguins[['bill_length_mm', 'bill_depth_mm']].plot(kind = 'box',
                                  title = 'Penguin flipper length (mm)',
                                  xlabel = 'Flipper length (mm)',
                                  grid = True)
```

```python
penguins['flipper_l'].plot(kind = 'hist',
                                  title = 'Penguin flipper length (mm)',
                                  xlabel = 'Flipper length (mm)',
                                  grid = True)
```
